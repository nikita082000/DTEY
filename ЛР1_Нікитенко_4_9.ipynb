Поелементний добуток — просто перемножує числа, що стоять на однакових місцях  в обох матрицях. Матриці мають бути однакового розміру.

Матричний добуток — перемножує рядки першої матриці на стовпці другої та додає отримані результати. Кількість стовпців першої матриці має дорівнювати кількості рядків другої.

### Завдання 28
Для матриці:

```text
A = [[4, 7],
     [2, 6]]
```

обчисліть:
- визначник;
- обернену матрицю.

Перевірте результат, помноживши початкову матрицю на обернену.

````python
# Код студента
A = np.array([
    [4, 7],
    [2, 6]
])

det = np.linalg.det(A)
inverse = np.linalg.inv(A)

print("Визначник:", det)
print("Обернена матриця:")
print(inverse)

print("Перевірка:")
print(A @ inverse)
````

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
# Код студента
A = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
])

mean = np.mean(A, axis=0)
std = np.std(A, axis=0)

Z = (A - mean) / std

print("Середні:", mean)
print("Стандартні відхилення:", std)
print(Z)
````

Створіть квадратну матрицю `5 × 5` із випадкових цілих чисел від `1` до `20`.

Виконайте:
1. знайдіть суму головної діагоналі;
2. знайдіть суму побічної діагоналі;
3. знайдіть максимальний елемент кожного рядка;
4. знайдіть мінімальний елемент кожного стовпця;
5. замініть усі елементи, менші за середнє значення всієї матриці, на `0`.

````python
# Код студента
A = np.random.randint(1, 21, size=(5, 5))

print("Матриця:")
print(A)

print("Сума головної діагоналі:", np.trace(A))
print("Сума побічної діагоналі:", np.trace(np.fliplr(A)))

print("Максимуми рядків:", np.max(A, axis=1))
print("Мінімуми стовпців:", np.min(A, axis=0))

mean = np.mean(A)

A[A < mean] = 0

print("Середнє:", mean)
print("Змінена матриця:")
print(A)
````

### Завдання 39
Створіть матрицю `4 × 4` із довільних цілих чисел.

Для кожного рядка знайдіть його норму та створіть нову матрицю, у якій кожен рядок нормалізований до одиничної довжини.

Спробуйте виконати завдання без циклів.

````python
# Код студента
A = np.arange(1, 17).reshape(4, 4)

norms = np.linalg.norm(
    A,
    axis=1,
    keepdims=True
)

normalized = A / norms

print("Початкова матриця:")
print(A)

print("Норми рядків:")
print(norms)

print("Нормалізована матриця:")
print(normalized)
````

Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
