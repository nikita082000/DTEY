```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 200)
y1 = np.sin(x)
y2 = np.cos(x)

fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(x, y1, linestyle='-', marker='o', markevery=20, label='sin(x)')
ax.plot(x, y2, linestyle='--', marker='s', markevery=20, label='cos(x)')

ax.set_title("Дві лінії")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.grid(True, alpha=0.3)
ax.legend()
plt.show()
```

```python
np.random.seed(1)
x = np.random.randn(200)
y = 1.2 * x + np.random.randn(200) / 2

c = np.hypot(x, y)  # “інтенсивність” для кольору

fig, ax = plt.subplots(figsize=(6, 3))
sc = ax.scatter(x, y, c=c, s=50, edgecolor='k')
ax.set_title("Scatter: колір за величиною")
ax.set_xlabel("x")
ax.set_ylabel("y")
plt.colorbar(sc, ax=ax, label="шкала кольору")
ax.grid(True, alpha=0.3)

plt.show()
```

```python
import matplotlib.pyplot as plt

cats = ["A", "B", "C", "D", "E", "F"]
vals = [12, 7, 13, 9, 11, 5]

fig, ax = plt.subplots(figsize=(6, 3))
bars = ax.bar(cats, vals)

ax.set_title("Bar chart із підписами")
ax.set_xlabel("Категорія")
ax.set_ylabel("Значення")
ax.grid(axis='y', alpha=0.3)

for b in bars:
    ax.text(b.get_x() + b.get_width()/2,
            b.get_height() + 0.3,
            f"{b.get_height():.0f}",
            ha="center", va="bottom")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 8, 200)
y1 = 3 + 0.5 * x + 0.2 * np.sin(2*x)
y2 = 1 + 0.3 * x + 0.2 * np.cos(2*x)

fig, ax = plt.subplots(figsize=(6, 3))
ax.fill_between(x, y1, y2, alpha=0.5)
ax.plot(x, (y1 + y2)/2, linewidth=2, label="Середня лінія")

ax.set_title("fill_between")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(5)

x = np.linspace(0, 2*np.pi, 200)
y = np.sin(x)
a = np.random.randn(300)

fig, axs = plt.subplots(2, 2, figsize=(8, 5))

# Line
axs[0, 0].plot(x, y)
axs[0, 0].set_title("Line")

# Scatter
axs[0, 1].scatter(np.random.randn(100), np.random.randn(100),
                  s=40, edgecolor='k')
axs[0, 1].set_title("Scatter")

# Bar
axs[1, 0].bar(["A", "B", "C"], [3, 7, 5])
axs[1, 0].set_title("Bar")

# Histogram
axs[1, 1].hist(a, bins=20)
axs[1, 1].set_title("Histogram")

# Сітка на всіх
for ax in axs.ravel():
    ax.grid(True, alpha=0.2)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.arange(0.0, 5.0, 0.01)
s = np.cos(2 * np.pi * t)

fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(t, s, linewidth=2)

ax.annotate("local max",
            xy=(2, 1),
            xytext=(3, 1.5),
            arrowprops=dict(facecolor="black", shrink=0.05))

ax.set_title("Анотація точки")
ax.set_ylim(-2, 2)
ax.grid(True, alpha=0.3)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(X) * np.cos(Y) + 0.5 * np.exp(-((X-1)**2 + (Y-1)**2))

fig, ax = plt.subplots(figsize=(6, 4))
contour = ax.contour(X, Y, Z, levels=20, cmap='viridis')
ax.clabel(contour, inline=True, fontsize=8)
ax.set_title("Contour plot Z(X,Y)")
ax.set_xlabel("X")
ax.set_ylabel("Y")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

fig = plt.figure(figsize=(7, 5))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='plasma', alpha=0.8)

ax.set_title("3D поверхня Z = sin(√(X² + Y²))")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec

np.random.seed(0)
t = np.linspace(0, 20, 300)
trend = 0.15 * t + np.sin(t) + np.random.normal(0, 0.3, len(t))

x = np.random.randn(400)
y = 0.8 * x + np.random.randn(400) / 2

fig = plt.figure(figsize=(10, 8))
gs = GridSpec(2, 2, figure=fig, height_ratios=[1.5, 1])

ax0 = fig.add_subplot(gs[0, :])
ax0.plot(t, trend, linewidth=2, color='tab:blue')
ax0.set_title("Тренд з часом")
ax0.grid(alpha=0.3)

ax1 = fig.add_subplot(gs[1, 0])
ax1.hist(np.random.normal(0, 1, 1000), bins=30, alpha=0.7, color='tab:orange')
ax1.set_title("Гістограма")

ax2 = fig.add_subplot(gs[1, 1])
ax2.scatter(x, y, alpha=0.6, s=20, color='tab:green')
ax2.set_title("Scatter plot")

fig.suptitle("Dashboard з GridSpec — 4 графіки", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

x = np.linspace(0, 2*np.pi, 400)
y = np.sin(x)

fig, ax = plt.subplots(figsize=(7, 4))
line, = ax.plot(x, y, linewidth=2, color='tab:blue')
point, = ax.plot([], [], 'ro', markersize=8)
text = ax.text(0.1, 0.9, '', transform=ax.transAxes, fontsize=12)

ax.set_xlim(0, 2*np.pi)
ax.set_ylim(-1.1, 1.1)
ax.grid(True, alpha=0.3)
ax.set_title("Анімація точки по sin(x)")

def animate(i):
    point.set_data([x[i]], [y[i]])
    text.set_text(f"x = {x[i]:.2f},  sin(x) = {y[i]:.3f}")
    return point, text

ani = FuncAnimation(fig, animate, frames=len(x), interval=30, blit=True)

plt.show()
```
